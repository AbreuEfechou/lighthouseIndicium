# Recomendações de Filmes

Neste notebook, abordaremos as recomendações de filmes com base na base de dados `desafio_indicium_imdb.csv`. Vamos explorar quais filmes recomendar e os fatores que influenciam o faturamento. Além disso, discutiremos a coluna `Overview` e sua relevância nas recomendações.

```python
# Importando bibliotecas necessárias
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configurando o estilo do seaborn
sns.set(style='whitegrid')
```

```python
# Carregando os dados
data_path = '../data/desafio_indicium_imdb.csv'
df = pd.read_csv(data_path)

# Exibindo as primeiras linhas do dataframe
df.head()
```

## Análise Inicial

Vamos realizar uma análise inicial dos dados para entender melhor as colunas disponíveis e suas características.

```python
# Informações gerais sobre o dataframe
df.info()

# Estatísticas descritivas
df.describe()
```

## Recomendações Baseadas em Fatores

Aqui, vamos explorar quais fatores podem influenciar as recomendações de filmes, como gênero, ano de lançamento e sinopse.

```python
# Analisando a distribuição de gêneros
genre_counts = df['Genre'].value_counts()
plt.figure(figsize=(12, 6))
sns.barplot(x=genre_counts.index, y=genre_counts.values)
plt.title('Distribuição de Gêneros de Filmes')
plt.xlabel('Gênero')
plt.ylabel('Número de Filmes')
plt.xticks(rotation=45)
plt.show()
```

## Análise da Coluna Overview

A coluna `Overview` pode fornecer insights valiosos sobre os filmes. Vamos analisar sua distribuição e como ela pode influenciar as recomendações.

```python
# Exibindo a distribuição do tamanho da coluna Overview
df['Overview_length'] = df['Overview'].apply(len)
plt.figure(figsize=(12, 6))
sns.histplot(df['Overview_length'], bins=30, kde=True)
plt.title('Distribuição do Tamanho da Coluna Overview')
plt.xlabel('Número de Caracteres')
plt.ylabel('Frequência')
plt.show()
```
